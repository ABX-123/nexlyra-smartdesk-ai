```python
#Task 46 - Check whether sla_target_hours matches the documented priority rule.

'''
if always
priority_reported = critical <=> sla_target_hours = 4
priority_reported = high <=> sla_target_hours = 12
priority_reported = medium <=> sla_target_hours = 24
priority_reported = low <=> sla_target_hours = 48
then
it matches the documented priority rule.
'''

#suppose it is true, then traverse the entire dataframe searching for a contradictory case
match = True

for priority, hours in zip(support_tickets['priority_reported'], support_tickets['sla_target_hours']):
    if priority == "Critical" and hours != 4:
        match = False
    elif priority == "High" and hours != 12:
        match = False
    elif priority == "Medium" and hours != 24:
        match = False
    elif priority == "Low" and hours != 48:
        match = False

    if match == False:
        break

if match == True:
    print("sla_target_hours matches the documented priority rule\n") #this is printed
elif match == False:
    print("sla_target_hours does not match the documented priority rule\n")

    
#in order to verify more, we can change 1 value in this way:
#support_tickets.at[2, 'sla_target_hours'] = 13
#then match becomes False
```
